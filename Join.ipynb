```python
import pandas as pd

# Tabel 1: Data Karyawan
df_karyawan = pd.DataFrame({
    'id_karyawan': [1, 2, 3, 4],
    'nama': ['Andi', 'Budi', 'Cici', 'Dodi']
})

# Tabel 2: Data Departemen
df_dept = pd.DataFrame({
    'id_karyawan': [1, 2, 3, 5],
    'departemen': ['IT', 'HR', 'Finance', 'Marketing']
})

# Melakukan Inner Join
hasil_inner = pd.merge(df_karyawan, df_dept, on='id_karyawan', how='inner')

print(hasil_inner)
```
